```python
#feature engineering (feature is column)
#quabtile(percentile)
import numpy as np
import pandas as pd

# Set the random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
## Create my DataFrame with Imbalanced Dataset.
class_0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2' : np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
class_0.head()
```

```python
class_1.head()
```

```python
df=pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
## upsampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minority_upsampled = resample(df_minority, replace=True, #Sample with replacement (means: A row that has already been selected can be selected again.)
         n_samples=len(df_majority), 
         random_state=42
         )
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```

```python
## Create my DataFrame with Imbalanced Dataset.
class_0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2' : np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

df = pd.concat([class_0, class_1]).reset_index(drop=True)

# Check the class distribution
print(df['target'].value_counts())
```

```python
## downsampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
df_majority_downsampled = resample(df_majority, replace=False, #Sample with replacement
         n_samples=len(df_minority), 
         random_state=42
         )
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([df_minority, df_majority_downsampled])
```

```python
df_downsampled.target.value_counts()
```

```python
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=1000,n_redundant=0,n_features=2,n_clusters_per_class=1,
                        weights=[0.90],random_state=12)
#x is any independent feature,y is my dependent feature(target/output)
#i have 1000 samples, 2 features,90% of the data belongs to one class and 10% belongs to the other class
```

```python
import pandas as pd
df1=pd.DataFrame(x,columns=['f1','f2'])
df2=pd.DataFrame(y,columns=['Target'])
final_df=pd.concat([df1,df2],axis=1)
final_df.head()
```

```python
final_df['Target'].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(final_df['f1'],final_df['f2'],c=final_df['Target'])
#interpolating technique used by SMOTE(see that below graph)
```

```python
from imblearn.over_sampling import SMOTE #(imblearn is pakage)
```

```python
#transform the dataset(SMOTE is a class)
oversample=SMOTE()
x,y=oversample.fit_resample(final_df[['f1','f2']],final_df['Target'])
```
