```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pf.groupby('OrderYear')['ShippingFee'].sum().plot(kind='line',marker='o',figsize=(8,4))
plt.title('Total Shipping Fee by year')
plt.xlabel('Year');
plt.ylabel('Shippinf Fee')
plt.grid(True)
plt.show()
```

```python
pf.groupby(pf['OrderDate'].dt.to_period('M')).size().plot(figsize=(10,4))
plt.title('Orders per Month')
plt.xlabel('Month'); plt.ylabel('Orders')
plt.show()
```

```python
pd.crosstab(pf['OrderStatus'], pf['PaymentMethod']).plot(kind='bar', stacked=True, figsize=(8,5))
plt.title('Order Status by Payment Method')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=pf, x='ShippingFee', y='rupees', hue='OrderStatus', palette='viridis')
plt.title('Shipping Fee vs Order Amount')
plt.xlabel('Shipping Fee')
plt.ylabel('Order Amount (rupees)')
plt.show()
```

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
plt.title('Gender Share')
plt.ylabel('')
plt.show()
```

```python
df['Gender'].value_counts().plot(kind='bar', color='steelblue', figsize=(8,4))
plt.title('Customers by Gender')
plt.xlabel('Gender'); plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
