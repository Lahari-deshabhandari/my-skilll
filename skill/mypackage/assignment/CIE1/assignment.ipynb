```python
arrays = [
    ["A", "A", "B", "B"],
    ["X", "Y", "X", "Y"]
]

index = pd.MultiIndex.from_arrays(
    arrays, names=["Category", "Subcategory"]
)

df = pd.DataFrame(
    {"Value": [10, 20, 30, 40]},
    index=index
)

print("Original DataFrame:")
print(df)

print("Select category A:")
print(df.loc["A"])

print("Select value at A, X:")
print(df.loc[("A", "X")])

print("Slice category A to B:")
print(df.loc["A":"B"])
```

```python
arrays = [['A','A','B','B'], ['X','Y','X','Y']]
index = pd.MultiIndex.from_arrays(arrays, names=('Category','SubCategory'))
df = pd.DataFrame(np.random.randint(1,10,(4,1)), index=index, columns=['Val'])
print(df.groupby(['Category','SubCategory']).sum())
```

```python
df = pd.DataFrame(
    np.random.randint(1, 11, size=(5, 3)),
    columns=["A", "B", "C"]
)

print("Original DataFrame:")
print(df)

def double(x):
    return x * 2

result = df.map(double)

print("After doubling:")
print(result)
```

```python
df = pd.DataFrame(
    np.random.randint(1, 11, size=(6, 3)),
    columns=["A", "B", "C"]
)

df["Sum"] = df.apply(
    lambda row: row["A"] + row["B"] + row["C"],
    axis=1
)

print(df)
```
