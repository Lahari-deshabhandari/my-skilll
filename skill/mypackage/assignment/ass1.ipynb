```python
import pandas as pd
```

```python
df=pd.DataFrame({
    "A":[2,3,5,6,7],
    "B":[3,8,6,4,5],
    "C":[1,6,5,3,7]

})
df["product"]=df["A"]*df["B"]
print(df)
```

```python
df=pd.DataFrame({
    "A":[2,3,5,6,7],
    "B":[3,8,6,4,5],
    "C":[1,6,5,3,7]

})
print(df)
print("\n Row wise sum")
print(df.sum(axis=1))
print("\n column wise sum")
print(df.sum(axis=0))                    #see carefully a,b,c are colunm here
```

```python
df=pd.DataFrame({
    "A":[11,np.nan,20,30,90],
    "B":[9,90,35,67,np.nan],
    "C":[45,89,67,8,6],
    "D":[4,89,56,89,8]                #here it delete all the row containing nan
})
print("\n",df)
df=df.dropna()
print(df)
```

```python
df=pd.DataFrame({
    "category":["A","G","A","F"],
    "value":[10,20,30,40]
})
print(df)
result=df.groupby("category") ["value"].agg(["sum","mean"])
print(result)
```

```python
df=pd.DataFrame({
    "product":["pen","Book","pencil","Bag"],
    "category":["stationary","education","stationary","bags"],
    "sales":[100,200,300,400]
})
print(df)
print(df.groupby("category")["sales"].sum())
```

```python
df1=pd.DataFrame({
    "ID":[1,2,3,4],
    "Name":["A","B","C","D"]
})
df2=pd.DataFrame({
    "ID":[1,2,3,2],
    "Marks":[80,90,76,89]
})
print(pd.merge(df1,df2,on="ID"))
```

```python
index = pd.MultiIndex.from_tuples([
    ("A", 1),
    ("A", 2),
    ("B", 1),
    ("B", 2)
], names=["Category", "Number"])

df = pd.DataFrame({
    "Value": [10, 20, 30, 40]
}, index=index)

print(df)

print("Category A:")
print(df.loc["A"])

print("A, 1:")
print(df.loc[("A", 1)])
```

```python

df = pd.DataFrame({
    "Category": ["A", "A", "B", "B"],
    "SubCategory": ["X", "Y", "X", "Y"],
    "Value": [10, 20, 30, 40]
})

df = df.set_index(["Category", "SubCategory"])

print(df)

print(df.groupby(level=["Category", "SubCategory"]).sum())
```

```python
df = pd.DataFrame(
    np.random.randint(1, 11, size=(5, 3)),
    columns=["A", "B", "C"]
)

print("Original DataFrame:")
print(df)

def double(x):
    return x * 2

result = df.map(double)

print("After doubling:")
print(result)
```

```python
df = pd.DataFrame(
    np.random.randint(1, 11, size=(6, 3)),
    columns=["A", "B", "C"]
)

df["Sum"] = df.apply(
    lambda row: row["A"] + row["B"] + row["C"],
    axis=1
)

print(df)
```
