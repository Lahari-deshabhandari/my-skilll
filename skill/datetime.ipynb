```python
df=pd.DataFrame({'name':['Tom','Luci','lahari'],
                 'DoB':['08-05-2002','04-10-2006','09-28-2006']})     
#if you run this line which converts date into DoB .....so here after
# all program must contain DoB in place og ['date']
df
```

```python
df.info()
```

```python
df['DoB']=pd.to_datetime(df['DoB'],format='%m-%d-%Y') #if you run this line which converts date into DoB .....so here after
# all program must contain DoB in place og ['date'] (this code is convertion of string DoB to datetime DoB)
```

```python
df['year']=df['date'].dt.year
df['month']=df['date'].dt.month
df['day']=df['date'].dt.day
df
```

```python
df['year of the week']=df['date'].dt.dayofweek
df['is leap year']=df['date'].dt.is_leap_year
df
```

```python
dw_mapping={
    0:'Monday',
    1:'tuesday',
    2:'Wednesday',
    3:'Thursday',
    4:'Friday',
    5:'Saturday',
    6:'Sunday'
}
df['day_of_week_name']=df['date'].dt.weekday.map(dw_mapping)
```
